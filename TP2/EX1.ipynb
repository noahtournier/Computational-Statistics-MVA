```python
import numpy as np 
import random
import matplotlib.pyplot as plt
```

```python
def sample_discrete(x, p):
    s = [0]
    n = len(p)
    for i in range(n):
        s.append(s[-1] + p[i])

    u = random.random()

    for i in range(1, n + 1):
        if s[i-1] <= u < s[i]:
            return x[i-1]

    return x[-1] 
```

```python
x = [0, 7, 3, 5]
p = [0.1, 0.4, 0.3, 0.2]

N = 100000
samples = [sample_discrete(x, p) for i in range(N)]
freq = [samples.count(x_i) / N for x_i in x]

pos = range(len(x))
width = 0.4
plt.bar([k - width/2 for k in pos], freq, width, label="Empirique")
plt.bar([k + width/2 for k in pos], p, width, label="Théorique")
plt.xticks(pos, x)
plt.xlabel("$x_i$")
plt.ylabel("Probabilité")
plt.legend()
plt.show()
```
