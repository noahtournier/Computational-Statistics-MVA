## Exercise 1 ##
We consider the parametric family of uniform distributions $\left\{ \mathcal{U}(\theta), \theta \ge 0 \right\}$ on $\mathbb{R}$, where $\mathcal{U}(\theta)$ has the following density with respect to Lebesgue measure : 
$$ p_\theta(x) = \frac{1}{\theta}\mathbb{1}_{[0,\theta]}(x)$$
Let $X_1, \dots, X_n$ be an i.i.d sample with density $p_\theta$. the goal of this exercise is to compare several estimators of $\theta$.  

1. Compute $\mathbb{E}_\theta(X_1)$ and deduce an estimator $\hat \theta_1$ of $\theta$ by the method of moments. 
2. Compute the quadratic risk of $\hat \theta_1$.
3. Compute the maximum likelihood estimator $\hat \theta_2$.
4. Compute the quadratic risk of $\hat \theta_2$.
5. Which estimator is preferable ?

---

#### 1. 
$X_1$ has density $p_\theta$ so : 
$$
\mathbb{E}_\theta(X_1) = \int_\mathbb{R} x p_\theta(x) dx = \int_0^\theta x\frac{1}{\theta}dx = \frac{1}{\theta}\left[ \frac{x^2}{2}\right]_0^\theta = \boxed{\frac{\theta}{2}}
$$
Meaning that $\hat \theta_1 = 2\mathbb{E}_\theta(X_1)$

$X_1, \dots, X_n$ are i.i.d. Using the method of moments we have :
$$
\boxed{\hat \theta_1 := \frac{2}{n}\sum_{i=1}^n X_i}
$$


#### 2. 
The quadratic risk of $\hat \theta_1$ is defined by :
$\mathbb{E}_\theta[(\hat\theta_1-\theta)^2]$

Meaning that the risk is equal to :

$$
\mathbb{E}_\theta[(\hat\theta_1-\theta)^2]
=\mathbb{E}_\theta[\hat\theta_1^2]
+\theta^2
-2\theta\mathbb{E}_\theta[\hat\theta_1]
$$

First, let's calculate the variance. By question 1 we have :

$$
\mathbb{E}_\theta[\hat\theta_1^2] = \frac{4}{n^2}\mathbb{E}_\theta\left[\sum_{i=1}^n \sum_{j=1}^n X_i X_j\right] = \frac{4}{n^2}\left[ \mathbb{E}_\theta\left[\sum_{i=1}^n X_i^2\right] + \mathbb{E}_\theta\left[\sum_{i=1}^n \sum_{j=1,\: j\ne i}^n X_i X_j\right]\right]
$$

Therefore, because $X_1, \dots, X_n$ are i.i.d. we obtain :
$$
\begin{aligned}
\mathbb{E}_\theta[\hat\theta_1^2] = \frac{4}{n^2}\left[ n\mathbb{E}_\theta\left[X_1^2\right] + \sum_{i=1}^n \sum_{j=1,\: j\ne i}^n \mathbb{E}_\theta [X_i] \mathbb{E}_\theta[X_j] \right] &= \frac{4}{n^2}\left[ n\frac{\theta^2}{3} + n(n-1)\frac{\theta^2}{4}\right]\\ &= \frac{4\theta^2}{3n} + \theta^2 - \frac{\theta^2}{n}\\ &= \frac{\theta^2}{3n}+\theta^2
\end{aligned}
$$

Moreover :
$$
2\theta\mathbb{E}_\theta[\hat\theta_1] = 2\theta\mathbb{E}_\theta\left[\frac{2}{n}\sum_{i=1}^n X_i\right] = 4\theta\mathbb{E}_\theta\left[ X_1\right] = 2\theta^2
$$

So : 
$$
\begin{aligned}
\mathbb{E}_\theta[(\hat\theta_1-\theta)^2]
&= \frac{\theta^2}{3n}+\theta^2 + \theta^2 - 2\theta^2\\ &= \boxed{\frac{\theta^2}{3n}}
\end{aligned}
$$


#### 3.

Since the $X_i$ are i.i.d., the likelihood function is : $L(\theta)=\prod_{i=1}^n p_\theta(x_i).$

Meaning that :

$$
\begin{aligned}
L(\theta)
&=\frac{1}{\theta^n}
\prod_{i=1}^n\mathbf 1_{[0,\theta]}(x_i)\\
&=\frac{1}{\theta^n}
\mathbf 1_{{\max(X_1,\ldots,X_n)\le \theta}}
\end{aligned}
$$

When $\max(X_1,\ldots,X_n) > \theta$ the likelihood is zero. Moreover, the likelihood is a decreasing function in $\theta$ so it is maximised for the smallest value of $\theta$ possible. This value is $\max(X_1,\ldots,X_n)$

So the maximum likelihood estimator is:

$$
\boxed{\hat\theta_2
=\max(X_1,\ldots,X_n)}
$$

#### 4.

We want to calculate the quadratic risk of the second estimator. 

First, as $X_1, \dots, X_n$ are i.i.d. we have :
$$
\mathbb{P}(\max(X_1,\ldots,X_n)\leq x)
=\mathbb{P}(X_1\leq x,\ldots,X_n\leq x) = \Pi_{i=1}^n \mathbb{P}(X_i \le x) = \mathbb{P}(X_1 \le x)^n = \boxed{\left( \frac{x}{\theta}\right)^n}
$$

In order to obtain the density and compute the expectation we just have to differentiate this formula:

$$
f_{\hat \theta_2}(x)
=\frac{nx^{n-1}}{\theta^n},
\qquad 0\leq x\leq\theta.
$$

So we have :

$$
\begin{aligned}
\mathbb{E}_\theta[\hat \theta_2]
=\int_0^\theta x\frac{nx^{n-1}}{\theta^n}dx
&=\frac{n}{\theta^n}\int_0^\theta x^ndx \\
&=\boxed{\frac{n}{n+1}\theta}
\end{aligned}
$$

And :

$$
\mathbb{E}_\theta[\hat \theta_2^2]
=\frac{n}{\theta^n}\int_0^\theta x^{n+1}dx
=\boxed{\frac{n}{n+2}\theta^2}
$$

Using the same formula as question 2 :

$$
\begin{aligned}
\mathbb{E}_\theta[(\hat\theta_2-\theta)^2]
&=\mathbb{E}_\theta[\hat\theta_2^2]
+\theta^2
-2\theta\mathbb{E}_\theta[\hat\theta_2]\\
&= \frac{n}{n+2}\theta^2 + \theta^2 - 2\frac{n}{n+1}\theta^2\\
&= \frac{n\theta^2(n+1) + (n+1)(n+2)\theta^2 -2n\theta^2(n+2)}{(n+1)(n+2)}\\
&= \frac{2\theta^2}{(n+1)(n+2)}
\end{aligned}
$$

The quadratic risk for this estimator is : 
$$\mathbb{E}_\theta[(\hat\theta_2-\theta)^2] = \boxed{\frac{2\theta^2}{(n+1)(n+2)}}$$

#### 5. 

```python
import numpy as np 
import matplotlib.pyplot as plt

def risk_theta1(theta, n):
    return theta**2/(3*n)

def risk_theta2(theta, n):
    return 2*theta**2/((n+1)*(n+2))

def compare_estimators(n_fixed=10, theta_max=100, n_max=300):

    theta = np.linspace(0.01, theta_max, 200)
    n = np.arange(1, n_max + 1)

    risk_1 = risk_theta1(theta, n_fixed)
    risk_2 = risk_theta2(theta, n_fixed)

    plt.figure(figsize=(8, 5))
    plt.plot(theta, risk_1, label=r'$R(\hat{\theta}_1)$', linewidth=2)
    plt.plot(theta, risk_2, label=r'$R(\hat{\theta}_2)$', linewidth=2)

    plt.xlabel(r'$\theta$')
    plt.ylabel('Quadratic risk')
    plt.title(f'Comparison of quadratic risks (n={n_fixed})')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    THETA, N = np.meshgrid(theta, n)

    risk_1_map = THETA**2 / (3 * N)
    risk_2_map = 2 * THETA**2 / ((N + 1) * (N + 2))
    better = np.where(risk_1_map < risk_2_map, 1, 2)

    plt.figure()
    plt.pcolormesh(THETA, N, better, vmin=1, vmax=2)
    plt.xlabel(r'$\theta$')
    cbar = plt.colorbar(ticks=[1, 2])
    cbar.ax.set_yticklabels([r'$\hat\theta$_1',r'$\hat\theta$_2'])
    plt.ylabel('n')
    plt.clim(1, 2)
    plt.show()

compare_estimators()
```

In order to know which estimator is the best, we just have to compare the quadratic risk of both. The second estimator is better than the first one when :

$$
\frac{2\theta^2}{(n+1)(n+2)}
<
\frac{\theta^2}{3n}.
$$

As $\theta$ is greater than 0 we have :

$$
6n<(n+1)(n+2).
$$
Which is equivalent to :
$$
6n<n^2+3n+2
$$

$$
(n-1)(n-2)>0
$$

$n$ is a positive integer, so for $n=1$ or $n=2$, both estimators have the same quadratic risk.
And for $n\ge 3$ the maximum likelihood estimator has a smaller quadratic risk.

Therefore:

$$
\boxed{
\begin{cases}
\mathbb{E}_\theta[(\hat\theta_2-\theta)^2]=\mathbb{E}_\theta[(\hat\theta_1-\theta)^2],
& n=1,2\\
\mathbb{E}_\theta[(\hat\theta_2-\theta)^2]<\mathbb{E}_\theta[(\hat\theta_1-\theta)^2],
& n\geq3
\end{cases}
}
$$
